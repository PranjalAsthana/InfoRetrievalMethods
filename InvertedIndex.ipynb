```python
import re
from collections import defaultdict
```

```python
documents = {
    1: "Java is a programming language.",
    2: "Python is a programming language.",
    3: "Java and Python are popular languages."
}
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

# # Common stop words
# stop_words = {
#     "is", "a", "the", "and", "are", "an", "in", "of", "to"
# }

# Use NLTK stop words for English
stop_words = set(stopwords.words('english'))
```

```python
inverted_index = defaultdict(set)

# Build inverted index
for doc_id, text in documents.items():

    # Lowercase
    text = text.lower()

    # Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)

    # Tokenization
    words = text.split()

    # Remove stop words
    words = [word for word in words if word not in stop_words]

    # Add to inverted index
    for word in words:
        inverted_index[word].add(doc_id)


# Display index
print("INVERTED INDEX")
print("=" * 30)

for word in sorted(inverted_index):
    print(f"{word:15} -> {sorted(inverted_index[word])}")
```

```python
# Search
def search(query):
    query = query.lower()
    return sorted(inverted_index.get(query, set()))


print("\nSEARCH")
print("=" * 30)

query = input("Enter search word: ")

result = search(query)

if result:
    print("Documents containing", query, ":", result)
else:
    print("Word not found.")
```
