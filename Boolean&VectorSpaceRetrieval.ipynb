```python
import pandas as pd
from contextlib import redirect_stdout
```

```python
import csv

data = [
    ["documents", "term1", "term2", "term3"],
    ["document1", "ice cream", "mango", "litchi"],
    ["document2", "hockey", "cricket", "sport"],
    ["document3", "litchi", "mango", "chocolate"],
    ["document4", "nice", "good", "cute"]
]

with open('dataset.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerows(data)

print("dataset.csv created successfully!")
```

```python
df = pd.read_csv("dataset.csv")
df
```

```python
terms = []
keys = []
vectors = {}
docdict = {}
dummy = []
def filter(docs, rows, cols):
    for i in range(rows):
        for j in range(cols):
            if(j==0):
                keys.append(docs.loc[i].iat[j])
            else:
                dummy.append(docs.loc[i].iat[j])
                if docs.loc[i].iat[j] not in terms:
                    terms.append(docs.loc[i].iat[j])

        copy = dummy.copy()
        docdict.update({docs.loc[i].iat[0]: copy})
        dummy.clear()
```

```python
def boolR(docdict, rows, cols):
    terms.sort()
    for i in docdict:
        for j in terms:
            if j in docdict[i]:
                dummy.append(1)
            else:
                dummy.append(0)
        copy = dummy.copy()
        vectors.update({i: copy})
        dummy.clear()
```

```python
def queryvector(query):
    qv = []
    for i in terms:
        if i in query:
            qv.append(1)
        else:
            qv.append(0)
    return qv


```

```python
def prediction(qv):
    dic = {}
    lis = []
    count = 0
    termlen = len(terms)

    for i in vectors:
        for t in range(termlen):
            if(qv[t]==vectors[i][t]):
                count+=1
        dic.update({i:count})
        count=0
    for i in dic:
        lis.append(dic[i])
    lis = sorted(lis, reverse=True)

    ans = ' '

    with open('output.txt','w') as f:
        with redirect_stdout(f):

            print("Ranking")
            for count, i in enumerate(lis):
                key = check(dic,i)
                if count==0:
                    ans=key

                print(key, "rank is", count+1)
                dic.pop(key)
            print(ans, "is the most relevant document for given query")
```

```python
def check(dic, val):
    for key, value in dic.items():
        if(val==value):
            return key
```

```python
def main():
    docs = df
    rows = len(docs)
    cols = len(docs.columns)

    filter(docs, rows, cols)
    boolR(docdict,rows,cols)

    print("Enter Query")
    query=input()
    query = query.split(' ')
    qve = queryvector(query)
    prediction(qve)

main()
```
