```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
text = "this is sample text, it will undergo preprocessing of various kinds, namely - stopword removal, tokenisation, and lemmatization. ! @ # $ % for no reason"
```

### Tokenisation

```python
print(word_tokenize(text))
```

### Stopword Removal

```python
sw = set(stopwords.words('english'))
filtered = [[word for word in text.split(' ') if word not in sw]]
filtered
```

### Lemmatization

```python
lemmatizer = WordNetLemmatizer()
lem = [lemmatizer.lemmatize(word) for word in text.split(' ')]
lem
```
